# VisualSeek — 03. CLIP Vision-Language Embeddings

This notebook demonstrates extracting 512-dimensional joint vision-language embeddings with OpenAI CLIP (ViT-B/32) and verifying cross-modal alignment.

```python
import sys
from pathlib import Path
import numpy as np

proj_root = Path("..").resolve()
if str(proj_root) not in sys.path:
    sys.path.insert(0, str(proj_root))

from src.models.clip_encoder import CLIPEncoder
from src.search.similarity import cosine_similarity_search
```

```python
# Initialize CLIP encoder
encoder = CLIPEncoder(model_name="ViT-B-32", pretrained="openai", device="auto")
print(f"Loaded {encoder.name}, Output Dim: {encoder.embedding_dim}")
```

```python
# Text-to-text semantic similarity check
texts = [
    "a fluffy white puppy",
    "a small cute white dog",
    "a dangerous mechanical submarine",
    "a golden retriever running outdoors"
]
text_embs = encoder.encode_text(texts, normalize=True)
sim_matrix = np.dot(text_embs, text_embs.T)

print("Cosine Similarity Matrix:")
for i, t in enumerate(texts):
    for j, t2 in enumerate(texts):
        print(f"{sim_matrix[i, j]:.3f} ", end="")
    print(f"<- '{t}'")
```
