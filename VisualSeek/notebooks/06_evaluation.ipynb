# VisualSeek — 06. Quantitative Evaluation & Benchmarking

This notebook loads the empirical benchmark results comparing Baseline (Color Histogram) vs CLIP (ViT-B/32) across Precision@K, Mean Reciprocal Rank (MRR), and retrieval latency.

```python
import sys
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

proj_root = Path("..").resolve()
bench_file = proj_root / "artifacts" / "benchmark_results.json"

with open(bench_file, "r") as f:
    data = json.load(f)

df = pd.DataFrame(data)
display(df[["model_name", "embedding_dim", "precision_at_1", "precision_at_5", "precision_at_10", "mrr", "mean_search_latency_ms"]])
```

```python
# Plot Precision@K comparison
ks = ["precision_at_1", "precision_at_5", "precision_at_10"]
labels = ["P@1", "P@5", "P@10"]

plt.figure(figsize=(8, 4))
for i, row in df.iterrows():
    scores = [row[k] * 100 for k in ks]
    plt.plot(labels, scores, marker="o", linewidth=2, label=row["model_name"])

plt.ylabel("Precision (%)")
plt.title("Precision@K Comparison on Stanford Dogs")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
