# VisualSeek — 01. Dataset Exploration & Discovery

This notebook demonstrates programmatic dataset discovery, validation, and exploratory data analysis (EDA) on the **Stanford Dogs Dataset** (20,580 images across 120 breeds).

```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

# Ensure project root is in path
proj_root = Path("..").resolve()
if str(proj_root) not in sys.path:
    sys.path.insert(0, str(proj_root))

from src.data.discovery import DatasetDiscovery
from src.data.loader import StanfordDogsDataset
```

```python
# 1. Run dynamic discovery
discovery = DatasetDiscovery()
loc = discovery.discover()
print(f"Dataset Discovered: {loc.name}")
print(f"Images Directory:   {loc.images_dir}")
print(f"Total Dog Breeds:   {loc.num_breeds}")
print(f"Total Images:       {loc.num_images:,}")
```

```python
# 2. Load and inspect sample records
dataset = StanfordDogsDataset(images_dir=loc.images_dir, limit=500, validate_on_load=True)
print(f"Loaded {len(dataset)} valid sampled images. Corrupt: {len(dataset.corrupt_files)}")

# Display sample images
fig, axes = plt.subplots(2, 4, figsize=(12, 6))
for i, ax in enumerate(axes.flat):
    rec = dataset[i]
    img = Image.open(rec.file_path)
    ax.imshow(img)
    ax.set_title(rec.breed, fontsize=9)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
