# VisualSeek — 05. t-SNE Non-Linear Manifold Learning

This notebook demonstrates projecting 512-dimensional CLIP vectors into 2D using scikit-learn t-SNE with PCA pre-reduction.

```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np

proj_root = Path("..").resolve()
if str(proj_root) not in sys.path:
    sys.path.insert(0, str(proj_root))

from src.embeddings.storage import EmbeddingStorage
from src.reduction.tsne import TSNEReducer
```

```python
storage = EmbeddingStorage(base_dir=proj_root / "artifacts")
embeddings, metadata, header = storage.load("default")

tsne = TSNEReducer(perplexity=30.0, max_iter=1000, random_state=42)
coords_2d = tsne.fit_transform(embeddings)
print(f"t-SNE coordinates computed. Shape: {coords_2d.shape}")

plt.figure(figsize=(9, 7))
plt.scatter(coords_2d[:, 0], coords_2d[:, 1], alpha=0.6, s=15, c="royalblue", edgecolors="none")
plt.title("t-SNE 2D Latent Manifold Projection of Dog Embeddings")
plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")
plt.grid(True, alpha=0.2)
plt.show()
```
