# VisualSeek — 04. Principal Component Analysis (PCA)

This notebook demonstrates linear dimensionality reduction with PCA, analyzing cumulative explained variance and 2D/3D projections.

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

proj_root = Path("..").resolve()
if str(proj_root) not in sys.path:
    sys.path.insert(0, str(proj_root))

from src.embeddings.storage import EmbeddingStorage
from src.reduction.pca import PCAReducer
```

```python
storage = EmbeddingStorage(base_dir=proj_root / "artifacts")
embeddings, metadata, header = storage.load("default")
print(f"Loaded {len(embeddings)} vectors with dimension {embeddings.shape[1]}.")

pca = PCAReducer(n_components=50, random_state=42)
transformed, var_ratio, cum_var = pca.fit_transform(embeddings)

# Scree plot
plt.figure(figsize=(10, 4))
plt.plot(range(1, 51), cum_var * 100, marker="o", markersize=3, label="Cumulative Variance (%)")
plt.bar(range(1, 51), var_ratio * 100, alpha=0.5, label="Individual Variance (%)")
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance (%)")
plt.title("PCA Explained Variance Curves")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
