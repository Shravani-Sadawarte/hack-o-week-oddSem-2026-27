# VisualSeek — 02. Color & Texture Histogram Baseline

This notebook explores the handcrafted baseline: 3D RGB and HSV color histograms with L2 normalization.

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

proj_root = Path("..").resolve()
if str(proj_root) not in sys.path:
    sys.path.insert(0, str(proj_root))

from src.data.loader import StanfordDogsDataset
from src.models.baseline import BaselineColorHistogramEncoder
from src.search.similarity import cosine_similarity_search
```

```python
dataset = StanfordDogsDataset(limit=100, seed=42)
encoder = BaselineColorHistogramEncoder(bins_rgb=4, bins_hsv=4)
print(f"Baseline Encoder: {encoder.name}, Dimensions: {encoder.embedding_dim}")

image_paths = [r.file_path for r in dataset.records]
embeddings = encoder.encode_image(image_paths, batch_size=32, normalize=True)
print(f"Extracted feature matrix shape: {embeddings.shape}")
```

```python
# Query sample image 0
query_vec = embeddings[0]
indices, scores = cosine_similarity_search(query_vec, embeddings, top_k=5)
print(f"Query: {dataset[0].breed} ({dataset[0].image_id})")
print("Top Retrieved Matches:")
for r, (idx, score) in enumerate(zip(indices, scores), 1):
    print(f"  #{r} {dataset[idx].breed} (Score: {score:.3f})")
```
